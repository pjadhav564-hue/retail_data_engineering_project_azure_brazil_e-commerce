```python
path = "/Volumes/e_commerce_brazil/e_com_bronze/bronze_clean"
sellers_df = read_csv(f"{path}/sellers")
print(sellers_df)
```

```python
#data cleaning and validation 

sellers_df = sellers_df.withColumnRenamed("sellerzip_code_prefix","seller_zip_code_prefix")\
    .withColumnRenamed("sellercity","seller_city")\
    .withColumnRenamed("sellertate","seller_state")
sellers_df = sellers_df.dropDuplicates(["seller_id"])
sellers_df = clean_column_names(sellers_df)
sellers_df = trim_space_col(sellers_df,["seller_id","seller_zip_code_prefix","seller_city","seller_state"])
sellers_df = cast_col(sellers_df,{"seller_zip_code_prefix":"integer"})
sellers_df = clean_string_column(sellers_df,lower_cols=["seller_city"],upper_cols=["seller_state"])
sellers_df = remove_null(sellers_df,["seller_id"])
sellers_df = handle_null(sellers_df,{"seller_zip_code_prefix":0,"seller_city":"unknown","seller_state":"unknown"})
sellers_df = ingest_ts(sellers_df)
display(sellers_df)
```
